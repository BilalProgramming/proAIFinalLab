```python
#PART A
import pandas as pd
df=pd.read_csv("/content/mystery_data.csv",header=None)
df.columns=["ID", "Math", "Gender", "StudyHours", "Science"]
df
# Justification for column :
#  1st column is unique so i give id.
# binary or categorical value e.g M/F so give Gender.
#  Math and Science Two numeric columns so i set it subject scores.
# StudyHours: look like time based hours

```

```python
# PART B
# Remove rows where 'StudyHours' is missing or invalid (e.g., null or non-numeric)
df = df.dropna(subset=["StudyHours"])
df

# Create a new column that calculates total score (Math + Science)
df["TotalScore"] = df["Math"] + df["Science"]
df["TotalScore"]




```

```python
# Print top 5 students based on total score
df.sort_values(by="TotalScore", ascending=False).head(5)
```

```python
# Filter students with StudyHours > 10
study_gt_10 = df[df["StudyHours"] > 10]
```

```python
import scipy.stats as stats
from scipy.stats import ttest_1samp
# Filter students with StudyHours > 10
study_gt_10 = df[df["StudyHours"] > 10]

# Perform one-sample t-test on Math scores
t_stat, p_value = ttest_1samp(study_gt_10["Math"], 70)

print(f"T-statistic: {t_stat}, P-value: {p_value}")
if p_value < 0.5:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Boxplot for Total Score by Gender
sns.boxplot(x="Gender", y="TotalScore", data=df)
plt.title("Total Score Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Score")
plt.show()
```
